# trynna count the number of fires in each year of hdf5_data

```python
import os

data_dir = "/home/miles/research/old_repo_FE_WSTS/hdf5_data"
test_years = ["2018", "2019", "2020", "2021"]

def count_files(directory):
    count = 0
    with os.scandir(directory) as entries:
        for entry in entries:
            if entry.is_file():
                count += 1
    return count

counts = {year: count_files(os.path.join(data_dir, year)) for year in test_years}
print(counts)
print("total:", sum(counts.values()))

# collect paths of every file named WildfireSpreadTS*
wsts_paths = []
for year in test_years:
    year_dir = os.path.join(data_dir, year)
    with os.scandir(year_dir) as entries:
        for entry in entries:
            if entry.is_file() and entry.name.startswith("WildfireSpreadTS"):
                wsts_paths.append(entry.path)

wsts_paths = sorted(wsts_paths)
wsts_paths
```
